# PIAR — interactive computational laboratory

Spring-coupled inverted pendulum (PIAR): $\ddot\theta + \gamma\dot\theta + \lambda\sin\theta\cos\theta - \beta\sin\theta = A\cos\Omega t$, with $\lambda=k/M$, $\beta=g/L$ and the control parameter $\mu = kL/(Mg)-1$ (natural units: $M=L=g=1$).

The notebook offers (i) the five conservative regimes of Table I with the four diagnostics of Fig. 2 (potential energy, kinetic energy, phase diagram, position), (ii) a smooth animation, and (iii) the six-panel dashboard for the damped–driven system. Every curve is drawn from a continuous representation of the solution (DOP853 dense output or cubic Hermite interpolation between RK4 steps), so trajectories look smooth at any zoom.

A browser version that needs no installation is provided as `web/piar_simulacion.html`.

## Smooth animation

The pendulum and its phase-space trajectory are rendered from the Hermite interpolant between integration steps (`speed` = natural time units per second of animation). Rendering the animation as JavaScript takes about a minute.

```python
from IPython.display import HTML
anim = animate(mu=-0.8, gamma=0.25, A=0.8, Omega=2/3, th0=0.0, om0=0.0, seconds=15, fps=30, speed=3)
HTML(anim.to_jshtml())
```

```python
try:
    import ipywidgets as W
    def _show(mu, gamma, A, Omega, th0, om0, n_periods):
        dashboard_figure(mu=mu, gamma=gamma, A=A, Omega=Omega, th0=th0, om0=om0, n_periods=n_periods)
        plt.show()
    W.interact_manual(_show,
        mu=W.FloatSlider(value=-0.8, min=-0.95, max=1.5, step=0.01, description='mu'),
        gamma=W.FloatSlider(value=0.25, min=0.0, max=1.0, step=0.01, description='gamma'),
        A=W.FloatSlider(value=0.8, min=0.0, max=1.6, step=0.005, description='A'),
        Omega=W.FloatSlider(value=2/3, min=0.3, max=1.6, step=0.01, description='Omega'),
        th0=W.FloatSlider(value=0.1, min=-3.14, max=3.14, step=0.01, description='theta0'),
        om0=W.FloatSlider(value=0.0, min=-3.0, max=3.0, step=0.01, description='omega0'),
        n_periods=W.IntSlider(value=1500, min=100, max=6000, step=100, description='periods'))
except ImportError:
    print('ipywidgets is not available: the default state is rendered')
    dashboard_figure(); plt.show()
```

## Suggested explorations

* **Critical softening.** With $\gamma = A = 0$, $\mu$ is moved from $-0.5$ to $+0.5$: the upright equilibrium changes from saddle to centre at $\mu = 0$, and two tilted saddles appear at $\cos\theta^* = 1/(1+\mu)$ (subcritical pitchfork).
* **Period doubling.** With $\mu=-0.8$, $\gamma=0.25$ and $\Omega=2/3$, $A$ is swept through $0.662$, $0.671$ and $0.6727$: the Poincaré section doubles its number of points (Feigenbaum cascade, accumulation at $A_\infty \approx 0.6732$).
* **Chaos.** At $A = 0.80$: strange attractor, broadband spectrum, $\lambda_1 \approx 0.096$.
* **Melnikov bound.** For $A < 0.512$ (with $\gamma=0.25$ and $\Omega=2/3$) no chaotic attractor is found.
